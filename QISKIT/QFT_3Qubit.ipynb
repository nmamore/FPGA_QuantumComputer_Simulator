```python
#initialization
import matplotlib.pyplot as plt
import numpy as np
from numpy import pi
import math
import time

# importing Qiskit
from qiskit import transpile
from qiskit import QuantumCircuit, ClassicalRegister, QuantumRegister
from qiskit_aer import Aer
from qiskit.circuit.library import Initialize

# import basic plot tools
from qiskit.visualization import plot_histogram
```

```python
#Init circuit for n = 2
n = 3
QFT = QuantumCircuit(n)

#Intialization data
in_dat = np.array([0.35+0.02j, -0.32+0j, 0.37 -0.02j, 0.35 - 0.04j, 0.3-0.02j, -0.34+0j, 0.37-0.04j,-0.41+0j])
in_norm = in_dat / np.linalg.norm(in_dat)
init = Initialize(in_norm)
QFT.append(init,[0, 1,2])
```

```python
#Quantum fourier transform on 3 qubits

#Q2 Operations
QFT.h(2)
QFT.cp(pi/2, 1, 2)
QFT.cp(pi/4, 0, 2)

#Q1 Operations
QFT.h(1)
QFT.cp(pi/2, 0, 1)

#Q0 operations
QFT.h(0)

#Place into standard format
QFT.swap(0,2)

#Measure result
QFT.measure_all()
QFT.draw()
```

```python
#Run simualtor for single sample to compare against FPGA
start_time = time.time()
qasm_sim = Aer.get_backend('qasm_simulator') #Init simulator
print(f"Optimized sim time: {time.time() - start_time:.10f}s")
result = qasm_sim.run(QFT, shots =1).result() #Capture result of runs
counts = result.get_counts()
plot_histogram(counts) #Plot
```

```python
#Run for 10000 Samples
start_time = time.time()
qasm_sim = Aer.get_backend('qasm_simulator') #Init simulator
print(f"Optimized sim time: {time.time() - start_time:.10f}s")
result = qasm_sim.run(QFT, shots =10000).result() #Capture result of runs
counts = result.get_counts()
plot_histogram(counts) #Plot
```
